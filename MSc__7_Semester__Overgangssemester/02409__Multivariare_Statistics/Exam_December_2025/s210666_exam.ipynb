```r
# Fit model M2
M2 <- lm(BSAAM ~ APMAM + APSAB, data = water)

# X-values for the 10th observation
new10 <- water[10, c("APMAM", "APSAB")]

# 95% prediction interval
pred10 <- predict(M2,
                  newdata = new10,
                  interval = "prediction",
                  level = 0.95)

pred10
```

```r
# Fitted value at obs 10
fit10 <- predict(M2, newdata = new10)

# Residual variance σ^2
sigma2 <- summary(M2)$sigma^2

# Leverage of obs 10
h10 <- hatvalues(M2)[10]

# Standard error for prediction interval
se_pred <- sqrt(sigma2 * (1 + h10))

# Degrees of freedom
df <- M2$df.residual          # 40

# t-quantile
tval <- qt(0.975, df)

# Interval endpoints
lower <- fit10 - tval * se_pred
upper <- fit10 + tval * se_pred

c(fit = fit10, se_pred = se_pred, lower = lower, upper = upper)
```

```r
# 2. Fit Model M2 (BSAAM ~ APMAM + APSAB)
M2 <- lm(BSAAM ~ APMAM + APSAB, data = water)

# 3. Extract the statistics for the APMAM parameter
# "coefficients" table gives Estimate and Std. Error
summ_M2 <- summary(M2)
apmam_stats <- summ_M2$coefficients["APMAM", ]

coef_val <- apmam_stats["Estimate"]     # The slope estimate
se_val   <- apmam_stats["Std. Error"]   # The standard error

# 4. Get the Degrees of Freedom and Critical T-value
# df = n - p (observations - parameters)
df_val <- M2$df.residual 
# qt(0.975, df) gives the 2-tailed t-value for 95% confidence
t_crit <- qt(0.975, df = df_val)

# 5. Print the values in the format of the question
cat("Formula: Estimate +/- t-value * Std.Error\n")
cat("-----------------------------------------\n")
cat("Estimate (APMAM):", coef_val, "\n")
cat("T-value (0.975, df=", df_val, "):", t_crit, "\n")
cat("Std. Error:", se_val, "\n")
cat("-----------------------------------------\n")
cat("Result matches option:\n")
cat(sprintf("%.3f +/- %f * %.3f", coef_val, t_crit, se_val), "\n")
```

```r
# Extract predictor and response variables
varX <- c("exciting", "calm", "independent", "sincere", "warm", "phyattr","strong") # assesment variables 
varY <- c("Years", "Serious", "ownPA") # reponses (crime )

# Correlation matrices (already standardized, Exx should be identity)
Exx = as.matrix(cor(MockJury[, c(varX)]))
Eyx = as.matrix(cor(MockJury[, c(varY)],MockJury[, c(varX)]))
Exy = as.matrix(cor(MockJury[, c(varX)],MockJury[, c(varY)]))
Eyy = as.matrix(cor(MockJury[, c(varY)])) 


# Inverse of Exx
invExx = solve(Exx)
invEyy = solve(Eyy)

Cancorr = geigen::geigen(Eyx%*%invExx%*%Exy,Eyy,symmetric = TRUE)
values = sort(Cancorr$values,decreasing = TRUE)

# H matrix
H = Eyx%*%invExx%*%Exy
E = Eyy - Eyx%*%invExx%*%Exy

invE = solve(E)
Ev <- eigen(invE%*%H)
var = Ev$values

# Eigenvalues, Proportion and Cumulative proportion of Variance:
varPC <- var/sum(var)
cumu = c(1:length(var))
for (i in 1:length(var)){
  cumu[i] = sum(varPC[1:i])
}

results <- data.frame("CanCor" = sqrt(values),"Squared CanCor" =
                        values,"eigenvalues"=var,"proportion"=varPC,
                      "cumulative" = cumu)

print(results) 
##      CanCor Squared.CanCor eigenvalues proportion cumulative
## 1 0.4086899     0.16702744  0.20051973 0.56825608  0.5682561
## 2 0.3453451     0.11926324  0.13541304 0.38374918  0.9520053
## 3 0.1290496     0.01665379  0.01693584 0.04799473  1.0000000
```

```r
# 1. Load the necessary library
library(MASS)

# 2. Load the Pima Indians Diabetes dataset
# We read directly from the source to ensure we have the exact same 768 observations.
url <- "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
col_names <- c("Pregnancies", "Glucose", "BloodPressure", "SkinThickness", 
               "Insulin", "BMI", "DiabetesPedigreeFunction", "Age", "Outcome")

data <- read.csv(url, header = FALSE, col.names = col_names)

# 3. Subset the variables as specified in the question
# Retain: Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin (plus Outcome)
data_subset <- data[, c("Pregnancies", "Glucose", "BloodPressure", 
                        "SkinThickness", "Insulin", "Outcome")]

# 4. Perform QDA with Equal Priors
# The 'prior' argument is set to c(0.5, 0.5) to specify equal probability 
# for Class 0 and Class 1, ignoring the actual imbalance in the data.
qda_model <- qda(Outcome ~ ., data = data_subset, prior = c(0.5, 0.5))

# 5. Make predictions on the same data
qda_pred <- predict(qda_model, data_subset)
predicted_class <- qda_pred$class

# 6. Create Confusion Matrix
# Rows = Actual, Columns = Predicted
conf_matrix <- table(Actual = data_subset$Outcome, Predicted = predicted_class)
print("Confusion Matrix:")
print(conf_matrix)

# 7. Calculate Sensitivity for Class 1
# Sensitivity = True Positives / Total Actual Positives
# In the matrix, [2,2] is True Positive (1 classified as 1) 
# and sum(conf_matrix[2, ]) is the Total Class 1 count.
TP <- conf_matrix["1", "1"]
total_class_1 <- sum(conf_matrix["1", ])

sensitivity <- TP / total_class_1

cat("\nTotal Class 1 observations:", total_class_1)
cat("\nTrue Positives (Correctly predicted Class 1):", TP)
cat("\nSensitivity:", round(sensitivity, 3))
```

```r
# 1. Define a function to check validity for a specific rho
check_matrix_validity <- function(rho) {
  # Construct the 4x4 matrix Sigma based on the problem description
  Sigma <- matrix(c(1,    rho^2, rho,   rho,
                    rho^2, 1,    rho,   rho,
                    rho,   rho,  1,     rho^2,
                    rho,   rho,  rho^2, 1), 
                  nrow = 4, ncol = 4, byrow = TRUE)
  
  # 2. Calculate Eigenvalues
  # A valid covariance matrix must have all non-negative eigenvalues
  ev <- eigen(Sigma)$values
  
  # Return TRUE if all eigenvalues are >= 0 (allowing for tiny floating point errors)
  return(all(ev > -1e-9))
}

# 3. Test a wide range of rho values (e.g., from -2 to 2)
rho_sequence <- seq(-2, 2, by = 0.1)
valid_results <- data.frame(rho = rho_sequence, 
                            IsValid = sapply(rho_sequence, check_matrix_validity))

# 4. Filter and print only the valid rho values
valid_rhos <- subset(valid_results, IsValid == TRUE)

print("Range of valid rho values:")
print(range(valid_rhos$rho))

# Optional: Inspect specific boundary cases from the options
cat("\nCheck specific values:")
cat("\nrho = 1.1 (Option > 1):", check_matrix_validity(1.1))
cat("\nrho = -1.1 (Option < -1):", check_matrix_validity(-1.1))
cat("\nrho = 0.5 (Inside -1 < rho < 1):", check_matrix_validity(0.5))
```
