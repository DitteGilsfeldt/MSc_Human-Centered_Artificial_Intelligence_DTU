## Miscellaneous

- **Sigmoid:** $\sigma(x) = (1 + e^{-x})^{-1}$
- **MLE:** $\hat{\theta}_{\text{MLE}} = \arg\max_\theta\, p(y|\theta)$ — maximise the likelihood
- **MAP:** $\hat{\theta}_{\text{MAP}} = \arg\max_\theta\, p(\theta|y)$ — maximise the posterior
- **Plug-in estimate:** use MAP or MLE directly instead of the full posterior distribution (overconfident — better to use posterior predictive)

### Uncertainty types
| Type | What it is | Captured by |
|------|-----------|------------|
| **Epistemic** | Uncertainty about model parameters | Posterior $p(\theta|y)$ |
| **Aleatoric** | Inherent noise in data — cannot be reduced | Likelihood $p(y|\theta)$ |
| **Both** | Total predictive uncertainty | Posterior predictive $p(y^*|y,x^*)$ |

### Metrics
- **MSE:** $\text{MSE} = \frac{1}{N}\sum_n (y_n - \hat{y}_n)^2$
- **ELPD:** $\text{ELPD} = \frac{1}{N}\sum_{n=1}^N \ln p(y_n^*|y, x^*)$ — like MSE but accounts for uncertainty

## Fundamental Probability Theory

| Rule | Formula |
|------|---------|
| **Product rule** | $p(a,b) = p(b|a)\,p(a)$ |
| **Sum rule / marginalisation** | $p(a) = \int p(a,b)\,db$ |
| **Conditional** | $p(a|b) = p(a,b)/p(b)$ |
| **Conditional independence** | $p(a,b|c) = p(a|c)\,p(b|c)$ — does NOT generally hold |
| **Bayes rule** | $p(a|b) = p(b|a)\,p(a)/p(b)$ |

### Expectations
$$\mathbb{E}[x] = \int x\,p(x)\,dx, \qquad \mathbb{V}[x] = \int (x - \mathbb{E}[x])^2\,p(x)\,dx, \qquad \mathbb{E}[f(x)] = \int f(x)\,p(x)\,dx$$

## Bayesian Inference

| Quantity | Notation | Meaning |
|----------|----------|---------|
| **Prior** | $p(\theta)$ | Belief about $\theta$ before seeing data |
| **Likelihood** | $p(y|\theta)$ | Probability of data given parameters |
| **Posterior** | $p(\theta|y)$ | Belief about $\theta$ after seeing data |
| **Evidence / marginal likelihood** | $p(y)$ | Probability of the data (normalising constant) |
| **Joint** | $p(\theta, y)$ | Probability of both parameters and data |
| **Posterior predictive** | $p(y^*|y,x,x^*)$ | Probability of new data $y^*$ given observations |

$$p(\theta|y) = \frac{p(y|\theta)\,p(\theta)}{p(y)}, \qquad p(y) = \int p(y|\theta)\,p(\theta)\,d\theta$$

## Distributions

### Gaussian $\mathcal{N}(x|\mu,\sigma^2)$ — continuous, $x\in\mathbb{R}$
$$p(x|\mu,\sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right), \qquad \mathbb{E}[x]=\mu, \quad \mathbb{V}[x]=\sigma^2$$

### Multivariate Gaussian $\mathcal{N}(x|\mu,\Sigma)$ — continuous, $x\in\mathbb{R}^D$
$$p(x|\mu,\Sigma) = \frac{1}{\sqrt{(2\pi)^D|\Sigma|}}\exp\!\left(-\frac{1}{2}(x-\mu)^T\Sigma^{-1}(x-\mu)\right), \qquad \mathbb{E}[x]=\mu, \quad \mathbb{V}[x]=\Sigma$$

### Bernoulli $\text{Ber}(x|\theta)$ — binary, $x\in\{0,1\}$
$$p(x|\theta) = \theta^x(1-\theta)^{1-x}, \qquad \mathbb{E}[x]=\theta, \quad \mathbb{V}[x]=\theta(1-\theta)$$

### Binomial $\text{Bin}(y|N,\theta)$ — $k$ successes from $N$ trials, $y\in\{0,\ldots,N\}$
$$p(y|N,\theta) = \binom{N}{y}\theta^y(1-\theta)^{N-y}, \qquad \mathbb{E}[y]=N\theta, \quad \mathbb{V}[y]=N\theta(1-\theta), \quad \hat{\theta}_{\text{MLE}}=y/N$$

Arises as $y=\sum_i x_i$ where $x_i\sim\text{Ber}(\theta)$ i.i.d.

### Beta $\text{Beta}(\theta|a,b)$ — continuous, $\theta\in[0,1]$
$$p(\theta|a,b) = \frac{1}{B(a,b)}\theta^{a-1}(1-\theta)^{b-1}, \qquad B(a,b)=\frac{\Gamma(a)\Gamma(b)}{\Gamma(a+b)}$$
$$\mathbb{E}[\theta]=\frac{a}{a+b}, \qquad \mathbb{V}[\theta]=\frac{ab}{(a+b)^2(a+b+1)}, \qquad \text{Uniform prior} = \text{Beta}(1,1)$$

### Conjugate pairs
A prior $p(\theta)$ is **conjugate** to a likelihood $p(y|\theta)$ if the posterior $p(\theta|y)$ is in the same family.

| Prior | Likelihood | Posterior | Update rule |
|-------|-----------|-----------|-------------|
| $\text{Beta}(a_0,b_0)$ | $\text{Bin}(N,\theta)$ | $\text{Beta}(a_0+y,\;b_0+N-y)$ | Add successes and failures |
| $\text{Gamma}(a_0,b_0)$ | $\text{Poisson}(\lambda)$ | $\text{Gamma}(a_0+\sum y_i,\;b_0+N)$ | Add counts and observations |
| $\mathcal{N}(m_0,S_0)$ | $\mathcal{N}(\Phi w, \sigma^2 I)$ | $\mathcal{N}(m,S)$ | Murphy eq. (3.37) |

Multivariate Gaussian is conjugate to itself.

## Techniques

### Grid Approximation
1. Define grid of parameter values $\{\theta_k\}$
2. Evaluate unnormalised posterior: $\tilde{\pi}_k = p(y,\theta_k|x)$
3. Normalise: $\pi_k = \tilde{\pi}_k / Z$ where $Z = \sum_k \tilde{\pi}_k$
4. Use $\{(\theta_k, \pi_k)\}$ as a discrete distribution for expectations, CIs, predictive

**Posterior predictive from grid:**
$$p(y^*|y,x^*) \approx \sum_k p(y^*|\theta_k, x^*)\,\pi_k$$

### Evidence Approximation (Type II MLE)
Pick hyperparameters that maximise the **marginal likelihood**:
$$\alpha^*,\beta^* = \arg\max_{\alpha,\beta}\ln p(y|\alpha,\beta)$$

Extension to Type II MAP: add a hyperprior $\ln p(\alpha,\beta)$ to the objective.

## Models

### Bayesian Linear Regression
Model: $y = \phi(x)^T w + \epsilon$, $\epsilon_n\sim\mathcal{N}(0,\sigma^2)$, noise precision $\beta = 1/\sigma^2$

$\phi(\cdot)$ is a basis function (e.g. polynomial), $\Phi$ is the design matrix with rows $\phi(x_n)^T$

| Quantity | Formula |
|----------|---------|
| **MLE** | $\hat{w}_{\text{MLE}} = (\Phi^T\Phi)^{-1}\Phi^T y$ |
| **Prior** | $p(w) = \mathcal{N}(w|m_0, S_0)$ |
| **Likelihood** | $p(y|w) = \mathcal{N}(y|\Phi w, \beta^{-1}I)$ |
| **Posterior covariance** | $S = (S_0^{-1} + \beta\Phi^T\Phi)^{-1}$ |
| **Posterior mean** | $m = S(S_0^{-1}m_0 + \beta\Phi^T y)$ |
| **Posterior** | $p(w|y) = \mathcal{N}(w|m, S)$ |
| **Evidence** | $p(y|\alpha,\beta) = \mathcal{N}(y|\Phi m_0,\; \beta^{-1}I + \Phi\alpha^{-1}I\Phi^T)$ |

**Isotropic prior** ($m_0=0$, $S_0=\alpha^{-1}I$):
$$S = (\alpha I + \beta\Phi^T\Phi)^{-1}, \qquad m = \beta S\Phi^T y$$

**Posterior predictive:**
$$p(f^*|y,x^*) = \mathcal{N}(f^*|m^T\phi^*,\;\phi^{*T}S\phi^*), \qquad p(y^*|y,x^*) = \mathcal{N}(y^*|m^T\phi^*,\;\phi^{*T}S\phi^* + \beta^{-1})$$

### GP Regression
Prior: $f\sim\mathcal{GP}(0,k)$, likelihood: $y_n = f(x_n)+\epsilon_n$, $\epsilon_n\sim\mathcal{N}(0,\sigma^2)$

$$\mu_{f^*} = k_*(K+\sigma^2I)^{-1}y, \qquad \sigma^2_{f^*} = k_{**} - k_*(K+\sigma^2I)^{-1}k_*^T$$
$$\log p(y|\theta) = -\frac{N}{2}\log(2\pi) - \frac{1}{2}\log|K+\sigma^2I| - \frac{1}{2}y^T(K+\sigma^2I)^{-1}y$$

**SE kernel:** $k(x,x') = \kappa^2\exp\!\left(-\frac{\|x-x'\|^2}{2\ell^2}\right)$, prior variance $= \kappa^2$

### Binary Classification (Logistic Regression)
Model: $y_n|w,x_n\sim\text{Ber}(\sigma(w^Tx_n))$, prior $w\sim\mathcal{N}(0,\lambda^{-1}I)$

**Log joint:** $\log p(y,w) = \sum_n[y_n\log\sigma(f_n)+(1-y_n)\log(1-\sigma(f_n))] - \frac{\lambda}{2}\|w\|^2$

**Hessian at MAP:** $\mathcal{H} = -X^TSX - \lambda I$, where $S_{nn}=\sigma(f_n)(1-\sigma(f_n))$

**Laplace:** $q(w)=\mathcal{N}(w|\hat{w}_{\text{MAP}}, S)$, $S=-\mathcal{H}^{-1}$

**Probit approximation:** $p(y^*=1|y,x^*) \approx \Phi\!\left(\frac{\mu_{f^*}}{\sqrt{8/\pi+\sigma^2_{f^*}}}\right)$

### Linear Gaussian Systems
System: $z=Ax+b+n_1$, $n_1\sim\mathcal{N}(0,\Sigma_1)$, prior $x\sim\mathcal{N}(\mu_x,\Sigma_x)$

**Direct conditional:** $p(z|x) = \mathcal{N}(z|Ax+b,\Sigma_1)$

**Reverse conditional** $p(x|z) = \mathcal{N}(x|m,S)$:
$$S^{-1} = \Sigma_x^{-1} + A^T\Sigma_1^{-1}A, \qquad m = S\left[A^T\Sigma_1^{-1}(z-b) + \Sigma_x^{-1}\mu_x\right]$$

**Marginalisation** — if $y=Wz+n_2$, $n_2\sim\mathcal{N}(0,\Sigma_2)$:
$$p(y) = \mathcal{N}(y|W\mu_z+b,\;\Sigma_2 + W\Sigma_z W^T)$$

### Beta-Binomial (conjugate)
$$p(\theta|y) = \text{Beta}(a_0+y,\;b_0+N-y), \qquad \mathbb{E}[\theta|y]=\frac{a_0+y}{a_0+b_0+N}$$

### Poisson-Gamma (conjugate)
$$p(\lambda|y) = \text{Gamma}\!\left(a_0+\sum_i y_i,\;b_0+N\right), \qquad \mathbb{E}[\lambda|y]=\frac{a_0+\sum y_i}{b_0+N}$$

### Variational Inference (Mean-Field)
$q(w) = \prod_i \mathcal{N}(w_i|m_i,v_i)$

$$\mathcal{L}[q] = \mathbb{E}_q[\log p(y,w)] + \mathcal{H}[q], \qquad \mathcal{H}[q] = \frac{1}{2}\sum_i\ln(2\pi e\,v_i)$$
$$\text{KL}[\mathcal{N}(m,v)\|\mathcal{N}(0,1)] = \frac{1}{2}(v+m^2-1-\ln v)$$

### MCMC — Metropolis
Proposal: $\theta^* = \theta^{(k-1)} + \epsilon$, $\epsilon\sim\mathcal{N}(0,\tau^2I)$

Accept if $\log U < \log p(\theta^*) - \log p(\theta^{(k-1)})$, where $U\sim\text{Uniform}(0,1)$

Normalising constant cancels in the ratio → only need unnormalised $\log\tilde{p}$

## Credibility Intervals

| Level | $z$ | Code |
|-------|-----|------|
| 80% | 1.282 | `norm.interval(0.80, loc=m, scale=sqrt(v))` |
| 90% | 1.645 | `norm.interval(0.90, loc=m, scale=sqrt(v))` |
| 95% | 1.960 | `norm.interval(0.95, loc=m, scale=sqrt(v))` |
| 99% | 2.576 | `norm.interval(0.99, loc=m, scale=sqrt(v))` |

**Warning:** $v$ is **variance** — always take $\sqrt{v}$ to get std before CI.

Tail probability: $p(w>c|y) = 1-\Phi\!\left(\frac{c-m}{\sqrt{v}}\right)$ → `1 - norm.cdf(c, loc=m, scale=sqrt(v))`

```python
# Quick numerical reference — run this to compute anything on the fly
import numpy as np
from scipy.stats import norm, beta as beta_dist, gamma as gamma_dist

# ── Credibility intervals ─────────────────────────────────────────────────────
m, v = -0.87, 0.41    # posterior mean and VARIANCE (not std)
std  = np.sqrt(v)
for pct, z in [(80,1.282),(90,1.645),(95,1.960),(99,2.576)]:
    lo, hi = norm.interval(pct/100, loc=m, scale=std)
    print(f"{pct}% CI: [{lo:.4f}, {hi:.4f}]  (z={z})")

# ── Tail probability ──────────────────────────────────────────────────────────
c = 0.0
print(f"p(w > {c}) = {1-norm.cdf(c, loc=m, scale=std):.4f}")

# ── Beta-Binomial ─────────────────────────────────────────────────────────────
a0,b0,N,y_obs = 1,1,115,4
a_post = a0+y_obs; b_post = b0+N-y_obs
lo,hi = beta_dist.interval(0.95, a=a_post, b=b_post)
print(f"Beta-Binomial posterior: Beta({a_post},{b_post}), mean={a_post/(a_post+b_post):.4f}, 95% CI=[{lo:.4f},{hi:.4f}]")

# ── Poisson-Gamma ─────────────────────────────────────────────────────────────
a0,b0 = 1,0.1; y_data=np.array([7,4,8,11,12])
a_pg = a0+np.sum(y_data); b_pg = b0+len(y_data)
lo,hi = gamma_dist.interval(0.95, a=a_pg, scale=1/b_pg)
print(f"Poisson-Gamma posterior: Gamma({a_pg},{b_pg}), mean={a_pg/b_pg:.4f}, 95% CI=[{lo:.4f},{hi:.4f}]")

# ── Probit approximation ──────────────────────────────────────────────────────
mu_f, var_f = 0.74, 2.91
p_probit = norm.cdf(mu_f / np.sqrt(8/np.pi + var_f))
print(f"Probit: p(y*=1|y,x*) = {p_probit:.4f}")

# ── Mean-field entropy ────────────────────────────────────────────────────────
v_vi = np.array([0.38,0.19,0.13,0.09])
print(f"Entropy H[q] = {0.5*np.sum(np.log(2*np.pi*np.e*v_vi)):.4f}")
```
