## **D1 | Show the posterior is Gaussian (completing the square)**

**Exam trigger:** "Show that the posterior $p(\mathbf{w}|\mathbf{y})$ is a Gaussian and identify its mean and covariance."

</div>

**Setup:** Bayesian linear regression:
$$p(\mathbf{w}) = \mathcal{N}(\mathbf{w}|\mathbf{0}, \alpha^{-1}\mathbf{I}), \qquad p(\mathbf{y}|\mathbf{w}) = \mathcal{N}(\mathbf{y}|\mathbf{\Phi}\mathbf{w}, \sigma^2\mathbf{I})$$

**Goal:** Show $p(\mathbf{w}|\mathbf{y}) = \mathcal{N}(\mathbf{w}|\mathbf{m}, \mathbf{S})$ and find $\mathbf{m}, \mathbf{S}$.

**Step 1: Log posterior (drop constants)**

By Bayes rule, $p(\mathbf{w}|\mathbf{y}) \propto p(\mathbf{y}|\mathbf{w})p(\mathbf{w})$, so:
$$\ln p(\mathbf{w}|\mathbf{y}) = \ln p(\mathbf{y}|\mathbf{w}) + \ln p(\mathbf{w}) + \text{const}$$

**Step 2: Expand, keep only w-dependent terms**

$$\ln p(\mathbf{y}|\mathbf{w}) = -\frac{1}{2\sigma^2}(\mathbf{y} - \mathbf{\Phi}\mathbf{w})^T(\mathbf{y} - \mathbf{\Phi}\mathbf{w}) + \text{const}$$

Expanding the quadratic:
$$= -\frac{1}{2\sigma^2}\left[\mathbf{y}^T\mathbf{y} - 2\mathbf{y}^T\mathbf{\Phi}\mathbf{w} + \mathbf{w}^T\mathbf{\Phi}^T\mathbf{\Phi}\mathbf{w}\right] + \text{const}$$

$$\ln p(\mathbf{w}) = -\frac{\alpha}{2}\mathbf{w}^T\mathbf{w} + \text{const}$$

**Step 3: Collect all w-dependent terms**

$$\ln p(\mathbf{w}|\mathbf{y}) = -\frac{1}{2}\mathbf{w}^T\underbrace{\left(\frac{1}{\sigma^2}\mathbf{\Phi}^T\mathbf{\Phi} + \alpha\mathbf{I}\right)}_{\mathbf{S}^{-1}}\mathbf{w} + \underbrace{\frac{1}{\sigma^2}\mathbf{y}^T\mathbf{\Phi}}_{\text{linear term}}\mathbf{w} + \text{const}$$

**Step 4: Match to canonical Gaussian form**

$\ln \mathcal{N}(\mathbf{w}|\mathbf{m}, \mathbf{S}) = -\frac{1}{2}\mathbf{w}^T\mathbf{S}^{-1}\mathbf{w} + \mathbf{m}^T\mathbf{S}^{-1}\mathbf{w} + \text{const}$

Matching the **second-order** coefficient:
$$\boxed{\mathbf{S}^{-1} = \alpha\mathbf{I} + \beta\mathbf{\Phi}^T\mathbf{\Phi}} \quad \Rightarrow \quad \mathbf{S} = (\alpha\mathbf{I} + \beta\mathbf{\Phi}^T\mathbf{\Phi})^{-1}$$

Matching the **first-order** coefficient:
$$\mathbf{m}^T\mathbf{S}^{-1} = \frac{1}{\sigma^2}\mathbf{y}^T\mathbf{\Phi} \quad \Rightarrow \quad \boxed{\mathbf{m} = \beta\mathbf{S}\mathbf{\Phi}^T\mathbf{y}}$$

**Conclusion:** $\ln p(\mathbf{w}|\mathbf{y})$ is quadratic in $\mathbf{w}$ $\Rightarrow$ posterior is Gaussian: $p(\mathbf{w}|\mathbf{y}) = \mathcal{N}(\mathbf{w}|\mathbf{m}, \mathbf{S})$

> **Key rule:** Quadratic log = Gaussian. Collect 2nd-order terms to get $\mathbf{S}^{-1}$, then 1st-order terms to get $\mathbf{m}$.

## **D2 | Show MAP = ridge regression**

**Exam trigger:** "Show that the MAP estimator under a Gaussian prior is equivalent to ridge regression."


</div>

**MAP definition:**
$$\hat{\mathbf{w}}_{\text{MAP}} = \arg\max_{\mathbf{w}} \left[\ln p(\mathbf{y}|\mathbf{w}) + \ln p(\mathbf{w})\right]$$

**Step 1: Write the objective**

$$= -\frac{1}{2\sigma^2}\|\mathbf{y} - \mathbf{\Phi}\mathbf{w}\|^2 - \frac{\alpha}{2}\|\mathbf{w}\|^2 + \text{const}$$

Maximising this is equivalent to minimising:
$$\mathcal{L}(\mathbf{w}) = \frac{1}{2\sigma^2}\|\mathbf{y} - \mathbf{\Phi}\mathbf{w}\|^2 + \frac{\alpha}{2}\|\mathbf{w}\|^2$$

Multiplying through by $2\sigma^2$:
$$\propto \|\mathbf{y} - \mathbf{\Phi}\mathbf{w}\|^2 + \underbrace{\alpha\sigma^2}_{\lambda}\|\mathbf{w}\|^2$$

**Step 2: This IS ridge regression** with $\lambda = \alpha/\beta = \alpha\sigma^2$.

**Step 3: Closed-form solution** - take gradient, set to zero:
$$\nabla_{\mathbf{w}}\mathcal{L} = -\mathbf{\Phi}^T(\mathbf{y} - \mathbf{\Phi}\mathbf{w}) + \alpha\sigma^2\mathbf{w} = 0$$

$$\boxed{\hat{\mathbf{w}}_{\text{MAP}} = (\mathbf{\Phi}^T\mathbf{\Phi} + \alpha\sigma^2\mathbf{I})^{-1}\mathbf{\Phi}^T\mathbf{y}}$$

**Limiting cases:**
- $\alpha \to 0$ (flat prior): $\hat{\mathbf{w}}_{\text{MAP}} \to \hat{\mathbf{w}}_{\text{MLE}} = (\mathbf{\Phi}^T\mathbf{\Phi})^{-1}\mathbf{\Phi}^T\mathbf{y}$
- $\alpha \to \infty$ (strong prior): $\hat{\mathbf{w}}_{\text{MAP}} \to \mathbf{0}$

> **Key insight:** Gaussian prior $\equiv$ $L_2$ regularisation. Laplace prior $\equiv$ $L_1$ (LASSO). Strength $\lambda = \alpha/\beta$.

## **D3 | Derive the log joint for a model**

**Exam trigger:** "Write down the log joint distribution $\ln p(\mathbf{y}, \mathbf{w})$ for the following model."

</div>


**Universal recipe:**
$$\ln p(\mathbf{y}, \mathbf{w}) = \underbrace{\sum_{n=1}^N \ln p(y_n|\mathbf{w})}_{\text{log likelihood}} + \underbrace{\ln p(\mathbf{w})}_{\text{log prior}}$$

### **Example 1: Logistic regression**
Model: $y_n|\mathbf{w} \sim \text{Ber}(\sigma(\mathbf{w}^T\phi_n))$, $\mathbf{w} \sim \mathcal{N}(\mathbf{0}, \alpha^{-1}\mathbf{I})$

$$\boxed{\ln p(\mathbf{y}, \mathbf{w}) = \sum_{n=1}^N \left[y_n \ln \sigma(f_n) + (1-y_n)\ln(1-\sigma(f_n))\right] - \frac{\alpha}{2}\|\mathbf{w}\|^2 + \text{const}}$$

where $f_n = \mathbf{w}^T\phi(\mathbf{x}_n)$.

### **Example 2: Poisson regression**
Model: $y_n|\mathbf{w} \sim \text{Poisson}(\exp(\mathbf{w}^T\mathbf{x}_n))$, $\mathbf{w} \sim \mathcal{N}(\mathbf{0}, \mathbf{I})$

Log likelihood of Poisson: $\ln p(y|\mu) = y\ln\mu - \mu - \ln(y!)$. Substituting $\mu_n = \exp(f_n)$:

$$\boxed{\ln p(\mathbf{y}, \mathbf{w}) = \sum_{n=1}^N \left[y_n \mathbf{w}^T\mathbf{x}_n - \exp(\mathbf{w}^T\mathbf{x}_n)\right] - \frac{1}{2}\|\mathbf{w}\|^2 + \text{const}}$$

### **Example 3: GP classification**
Model: $y_n|f_n \sim \text{Ber}(\sigma(f_n))$, $\mathbf{f} \sim \mathcal{N}(\mathbf{0}, \mathbf{K})$

$$\boxed{\ln p(\mathbf{y}, \mathbf{f}) = \sum_{n=1}^N \left[y_n \ln \sigma(f_n) + (1-y_n)\ln(1-\sigma(f_n))\right] - \frac{1}{2}\mathbf{f}^T\mathbf{K}^{-1}\mathbf{f} + \text{const}}$$

> **Tip:** Always: product rule $\to$ conditional independence $\to$ split into sum over $n$.

## **D4 | CAVI: Show $q^*(\mu)$ is Gaussian**

**Exam trigger:** "Using CAVI, show that the optimal variational factor $q^*(\mu)$ is Gaussian and identify its parameters."

</div>


**Setup:** $p(\mathcal{D}|\mu,\tau) = \prod_n \mathcal{N}(x_n|\mu,\tau^{-1})$, $p(\mu|\tau) = \mathcal{N}(\mu|\mu_0,(\lambda_0\tau)^{-1})$, $p(\tau) = \text{Gamma}(\tau|a_0,b_0)$

Variational family: $q(\mu,\tau) = q(\mu)q(\tau)$

**CAVI update rule:**
$$\ln q^*(\mu) = \mathbb{E}_{q(\tau)}[\ln p(\mathcal{D}, \mu, \tau)] + \text{const}$$

**Step 1: Extract mu-dependent terms from the log joint**

The $\mu$-dependent part of $\ln p(\mathcal{D},\mu,\tau)$ is:
$$\ni -\tau\left[-\mu(N\bar{x} + \lambda_0\mu_0) + \frac{1}{2}\mu^2(N + \lambda_0)\right]$$

where $N\bar{x} = \sum_n x_n$.

**Step 2: Take expectation over q(tau), absorb non-mu terms into constant**

$$\ln q^*(\mu) = -\frac{1}{2}\mu^2(N + \lambda_0)\mathbb{E}[\tau] + \mu(N\bar{x} + \lambda_0\mu_0)\mathbb{E}[\tau] + \text{const}$$

**Step 3: Match to** $\ln \mathcal{N}(\mu|m,v) = -\frac{1}{2v}\mu^2 + \frac{m}{v}\mu + \text{const}$

Matching 2nd-order:
$$\boxed{v = \frac{1}{(N + \lambda_0)\mathbb{E}[\tau]}}$$

Matching 1st-order, then substituting $v$:
$$\boxed{m = \frac{N\bar{x} + \lambda_0\mu_0}{N + \lambda_0}}$$

**Conclusion:** Log of $q^*(\mu)$ is quadratic in $\mu$ $\Rightarrow$

$$q^*(\mu) = \mathcal{N}\!\left(\mu\;\Big|\;\frac{N\bar{x} + \lambda_0\mu_0}{N + \lambda_0},\; \frac{1}{(N+\lambda_0)\mathbb{E}_{q(\tau)}[\tau]}\right)$$

> **Note:** $\mathbb{E}[\tau] = a/b$ for Gamma$(a,b)$. The mean $m$ does NOT depend on $\mathbb{E}[\tau]$ - it cancels!

## **D5 | Derive the ELBO for a model**

**Exam trigger:** "Derive the evidence lower bound (ELBO) for the following model."

</div>



**ELBO definition - always start here:**
$$\mathcal{L}[q] = \mathbb{E}_{q(\mathbf{w})}[\ln p(\mathbf{y}, \mathbf{w})] - \mathbb{E}_{q(\mathbf{w})}[\ln q(\mathbf{w})] = \mathbb{E}_q[\ln p(\mathbf{y}|\mathbf{w})] + \mathbb{E}_q[\ln p(\mathbf{w})] + \mathcal{H}[q]$$

### ELBO for Bayesian linear regression, mean-field $q(\mathbf{w}) = \prod_i \mathcal{N}(w_i|m_i,v_i)$

**Term 1 - Expected log likelihood:**

$$\mathbb{E}_q[\ln p(\mathbf{y}|\mathbf{w})] = \sum_n \ln \mathcal{N}(y_n|\mathbf{m}^T\phi_n, \sigma^2) - \frac{1}{2\sigma^2}\sum_i v_i \sum_n x_{ni}^2$$

Key step: $\mathbb{E}[w_iw_j] = m_im_j$ for $i\neq j$ (independence), $\mathbb{E}[w_i^2] = m_i^2 + v_i$ for $i=j$. This produces the extra $-\frac{v_i}{2\sigma^2}$ variance correction term.

**Term 2 - Expected log prior:**

$$\mathbb{E}_q[\ln \mathcal{N}(\mathbf{w}|\mathbf{0}, \kappa^2\mathbf{I})] = \sum_i \ln \mathcal{N}(m_i|0, \kappa^2) - \frac{1}{2\kappa^2}\sum_i v_i$$

Key step: $\mathbb{E}_q[w_i^2] = m_i^2 + v_i$, so the variance $v_i$ contributes an extra correction $-v_i/(2\kappa^2)$.

**Term 3 - Entropy:**
$$\mathcal{H}[q] = \frac{1}{2}\sum_i \ln(2\pi e\, v_i)$$

**Full ELBO:**
$$\boxed{\mathcal{L} = \sum_n \ln\mathcal{N}(y_n|\mathbf{m}^T\phi_n, \sigma^2) - \frac{\sum_i v_i\sum_n x_{ni}^2}{2\sigma^2} + \sum_i \ln\mathcal{N}(m_i|0,\kappa^2) - \frac{\sum_i v_i}{2\kappa^2} + \frac{1}{2}\sum_i\ln(2\pi e\,v_i)}$$

Why ELBO lower bounds ln p(y)

$$\ln p(\mathbf{y}) = \text{KL}[q\|p(\cdot|\mathbf{y})] + \mathcal{L}[q]$$

Since $\text{KL} \geq 0$ always: $\ln p(\mathbf{y}) \geq \mathcal{L}[q]$. Maximising $\mathcal{L}$ $\equiv$ minimising KL (since $\ln p(\mathbf{y})$ is constant wrt. $q$).

## **D6 | Linear Gaussian chain: $p(y)$, $p(y|z1)$, $p(z1|y)$**

**Exam trigger:** "Determine $p(y)$, $p(y|z1)$, and $p(z1|y)$ for the following Gaussian system." (Assignment 1 Part 2 pattern.)

</div>



**Setup:**
$$z_1 \sim \mathcal{N}(0, v\mathbf{I}), \quad z_2|z_1 \sim \mathcal{N}(z_1, v\mathbf{I}), \quad y|z_2 \sim \mathcal{N}(\mathbf{a}^T z_2, \sigma^2)$$

**Fundamental rules (memorise):** For $y = Ax + \epsilon$, $x \sim \mathcal{N}(\mu_x,\Sigma_x)$, $\epsilon \sim \mathcal{N}(0,\Sigma_\epsilon)$ independent:
- $\mathbb{E}[y] = A\mu_x$
- $\text{Var}[y] = A\Sigma_x A^T + \Sigma_\epsilon$
- $\text{Cov}(x,y) = \Sigma_x A^T$

**Finding p(z2)**

Write $z_2 = z_1 + \epsilon_2$, $z_1 \perp \epsilon_2$:
$$z_2 \sim \mathcal{N}(0, 2v\mathbf{I})$$

**Finding p(y)**

Write $y = \mathbf{a}^T z_2 + \epsilon_y$, $z_2 \perp \epsilon_y$:
$$\mathbb{E}[y] = 0, \quad \text{Var}[y] = \mathbf{a}^T(2v\mathbf{I})\mathbf{a} + \sigma^2 = 2v\|\mathbf{a}\|^2 + \sigma^2$$
$$\boxed{p(y) = \mathcal{N}(y\,|\,0,\; 2v\mathbf{a}^T\mathbf{a} + \sigma^2)}$$

**Finding p(y|z1)**

Condition on $z_1$. Write $y = \mathbf{a}^T z_1 + \mathbf{a}^T\epsilon_2 + \epsilon_y$. Given $z_1$, only $\mathbf{a}^T\epsilon_2 + \epsilon_y$ is random:
$$\mathbb{E}[y|z_1] = \mathbf{a}^T z_1, \quad \text{Var}[y|z_1] = v\|\mathbf{a}\|^2 + \sigma^2$$
$$\boxed{p(y|z_1) = \mathcal{N}(y\,|\,\mathbf{a}^Tz_1,\; v\|\mathbf{a}\|^2 + \sigma^2)}$$

**Finding p(z1|y) - Gaussian conditioning**

**Step 1:** Find the joint $(z_1, y)$:

$$\text{Cov}(z_1, y) = \text{Cov}(z_1, \mathbf{a}^Tz_1) = \text{Var}(z_1)\mathbf{a} = v\mathbf{a}$$

$$\begin{bmatrix}z_1\\y\end{bmatrix} \sim \mathcal{N}\!\left(\mathbf{0},\; \begin{bmatrix}v\mathbf{I} & v\mathbf{a} \\ v\mathbf{a}^T & 2v\|\mathbf{a}\|^2 + \sigma^2\end{bmatrix}\right)$$

**Step 2:** Apply conditioning formula $p(y_1|y_2) = \mathcal{N}(\mu_1 + \Sigma_{12}\Sigma_{22}^{-1}(y_2-\mu_2),\; \Sigma_{11} - \Sigma_{12}\Sigma_{22}^{-1}\Sigma_{21})$:

$$\mu_{z_1|y} = \frac{v\mathbf{a}}{2v\|\mathbf{a}\|^2 + \sigma^2}\,y, \qquad \Sigma_{z_1|y} = v\mathbf{I} - \frac{v^2\mathbf{a}\mathbf{a}^T}{2v\|\mathbf{a}\|^2 + \sigma^2}$$

$$\boxed{p(z_1|y) = \mathcal{N}\!\left(z_1\;\Big|\;\frac{v\mathbf{a}\,y}{2v\|\mathbf{a}\|^2+\sigma^2},\; v\mathbf{I} - \frac{v^2\mathbf{a}\mathbf{a}^T}{2v\|\mathbf{a}\|^2+\sigma^2}\right)}$$

## **D7 | Show Beta-Binomial posterior update (conjugacy)**

**Exam trigger:** "Show that the posterior p(theta|y) is a Beta distribution and identify its parameters."

</div>



**Setup:** $y|\theta \sim \text{Bin}(N, \theta)$, $\theta \sim \text{Beta}(a_0, b_0)$

**Step 1:** Write log posterior
$$\ln p(\theta|y) = \ln p(y|\theta) + \ln p(\theta) + \text{const}$$

**Step 2:** Expand, keep only theta-dependent terms
$$\ln p(y|\theta) = y\ln\theta + (N-y)\ln(1-\theta) + \text{const}$$
$$\ln p(\theta) = (a_0-1)\ln\theta + (b_0-1)\ln(1-\theta) + \text{const}$$

**Step 3:** Combine
$$\ln p(\theta|y) = (y + a_0 - 1)\ln\theta + (N - y + b_0 - 1)\ln(1-\theta) + \text{const}$$

**Step 4:** Match to $\ln\text{Beta}(a,b) = (a-1)\ln\theta + (b-1)\ln(1-\theta) + \text{const}$:
$$a = a_0 + y, \qquad b = b_0 + N - y$$

$$\boxed{p(\theta|y) = \text{Beta}(\theta|a_0 + y,\; b_0 + N - y)}$$

**Posterior mean:** $\hat{\theta}_{\text{Bayes}} = \frac{a_0 + y}{a_0 + b_0 + N}$

As $N \to \infty$: posterior mean $\to y/N$ (MLE). As $N \to 0$: posterior mean $\to a_0/(a_0+b_0)$ (prior mean).

## D8 | Laplace approximation - full worked example

**Exam trigger:** "Compute the Laplace approximation for the posterior $p(\mathbf{w}|\mathbf{y})$."

</div>



**Setup:** Logistic regression: $y_n \sim \text{Ber}(\sigma(\mathbf{w}^T\phi_n))$, $\mathbf{w} \sim \mathcal{N}(\mathbf{0}, \alpha^{-1}\mathbf{I})$

Goal: $q(\mathbf{w}) = \mathcal{N}(\mathbf{w}|\hat{\mathbf{w}}, \mathbf{S}) \approx p(\mathbf{w}|\mathbf{y})$

**Step 1: Find MAP**
$$\hat{\mathbf{w}} = \arg\max_{\mathbf{w}} \left[\sum_n y_n\ln\sigma(f_n) + (1-y_n)\ln(1-\sigma(f_n)) - \frac{\alpha}{2}\|\mathbf{w}\|^2\right]$$

Use `scipy.optimize.minimize` on the negative log joint in practice.

**Step 2: Compute Hessian at MAP**

Gradient:
$$\nabla_{\mathbf{w}}\ln p(\mathbf{y}, \mathbf{w}) = \mathbf{\Phi}^T(\mathbf{y} - \boldsymbol{\sigma}) - \alpha\mathbf{w}$$

Hessian:
$$\mathbf{H} = -\mathbf{\Phi}^T\mathbf{W}\mathbf{\Phi} - \alpha\mathbf{I}$$

where $\mathbf{W} = \text{diag}(\sigma(f_n)(1-\sigma(f_n)))_n$. Note $\mathbf{H}$ is negative definite at a maximum.

**Step 3: Set covariance = negative inverse Hessian**
$$\boxed{\mathbf{S} = (-\mathbf{H})^{-1} = (\mathbf{\Phi}^T\mathbf{W}\mathbf{\Phi} + \alpha\mathbf{I})^{-1}}$$

**Why it works:** Taylor expand $\ln p(\mathbf{y},\mathbf{w})$ around $\hat{\mathbf{w}}$:
$$\ln p(\mathbf{y}, \mathbf{w}) \approx \text{const} + \underbrace{\nabla|_{\hat{\mathbf{w}}}^T}_{=0}(\mathbf{w}-\hat{\mathbf{w}}) - \frac{1}{2}(\mathbf{w}-\hat{\mathbf{w}})^T(-\mathbf{H})(\mathbf{w}-\hat{\mathbf{w}})$$

This is the log of $\mathcal{N}(\mathbf{w}|\hat{\mathbf{w}}, (-\mathbf{H})^{-1})$.

**Posterior predictive (classification):**
$$p(y^*=1|\mathbf{y}, \mathbf{x}^*) \approx \Phi\!\left(\frac{\mu_{f^*}}{\sqrt{8/\pi + \sigma^2_{f^*}}}\right), \quad \mu_{f^*} = \hat{\mathbf{w}}^T\phi^*, \quad \sigma^2_{f^*} = (\phi^*)^T\mathbf{S}\phi^*$$

## **D9 | GP posterior predictive derivation**

**Exam trigger:** "Derive the posterior predictive distribution p(f*|y, x*) for GP regression."
    
</div>



**Setup:** $y_n = f(x_n) + \epsilon_n$, $\epsilon_n \sim \mathcal{N}(0,\sigma^2)$, $f \sim \mathcal{GP}(0, k(x,x'))$

**Step 1: Joint prior over f and f***

Any finite set of GP values is jointly Gaussian:
$$\begin{bmatrix}\mathbf{f}\\f^*\end{bmatrix} \sim \mathcal{N}\!\left(\mathbf{0},\; \begin{bmatrix}\mathbf{K} & \mathbf{k}_* \\ \mathbf{k}_*^T & k_{**}\end{bmatrix}\right)$$

where $\mathbf{K}_{nm} = k(x_n,x_m)$, $\mathbf{k}_{*n} = k(x^*,x_n)$, $k_{**} = k(x^*,x^*)$.

**Step 2: Joint of observations y and f***

The likelihood adds noise to training observations only:
$$\begin{bmatrix}\mathbf{y}\\f^*\end{bmatrix} \sim \mathcal{N}\!\left(\mathbf{0},\; \begin{bmatrix}\mathbf{K} + \sigma^2\mathbf{I} & \mathbf{k}_* \\ \mathbf{k}_*^T & k_{**}\end{bmatrix}\right)$$

**Step 3: Apply Gaussian conditioning formula**

Let $\mathbf{C} = \mathbf{K} + \sigma^2\mathbf{I}$:
$$\mu_{f^*|\mathbf{y}} = \mathbf{k}_*^T\mathbf{C}^{-1}\mathbf{y}$$
$$\sigma^2_{f^*|\mathbf{y}} = k_{**} - \mathbf{k}_*^T\mathbf{C}^{-1}\mathbf{k}_*$$

$$\boxed{p(f^*|\mathbf{y},x^*) = \mathcal{N}\!\left(f^*\;\Big|\;\mathbf{k}_*^T(\mathbf{K}+\sigma^2\mathbf{I})^{-1}\mathbf{y},\; k_{**} - \mathbf{k}_*^T(\mathbf{K}+\sigma^2\mathbf{I})^{-1}\mathbf{k}_*\right)}$$

**For y* (noisy observation):** Add $\sigma^2$:
$$\boxed{p(y^*|\mathbf{y},x^*) = \mathcal{N}\!\left(y^*\;\Big|\;\mathbf{k}_*^T\mathbf{C}^{-1}\mathbf{y},\; k_{**} - \mathbf{k}_*^T\mathbf{C}^{-1}\mathbf{k}_* + \sigma^2\right)}$$

**Interpretation:**
- Posterior mean: weighted sum of training observations; weights $\mathbf{C}^{-1}\mathbf{k}_*$ decay with distance from $x^*$
- Posterior variance: prior variance $k_{**}$ reduced by data term $\mathbf{k}_*^T\mathbf{C}^{-1}\mathbf{k}_*$; goes to prior variance far from data
- $f^*$ = latent (epistemic only); $y^*$ adds aleatoric $\sigma^2$

## **D10 | MH acceptance probability - why the normaliser cancels**

**Exam trigger:** "Show that the MH algorithm does not require the normalising constant $p(\mathbf{y})$."

</div>



**MH acceptance ratio:**
$$r_k = \frac{p_t(\theta^*)\, q(\theta^{k-1}|\theta^*)}{p_t(\theta^{k-1})\, q(\theta^*|\theta^{k-1})}$$

**Target is a posterior:** $p_t(\theta) = p(\theta|\mathbf{y}) = \frac{p(\mathbf{y}|\theta)p(\theta)}{p(\mathbf{y})}$

**Substituting:**
$$r_k = \frac{\dfrac{p(\mathbf{y}|\theta^*)p(\theta^*)}{p(\mathbf{y})}\, q(\theta^{k-1}|\theta^*)}{\dfrac{p(\mathbf{y}|\theta^{k-1})p(\theta^{k-1})}{p(\mathbf{y})}\, q(\theta^*|\theta^{k-1})}$$

**The normalising constant $p(\mathbf{y})$ cancels:**
$$\boxed{r_k = \frac{p(\mathbf{y},\theta^*)\,q(\theta^{k-1}|\theta^*)}{p(\mathbf{y},\theta^{k-1})\,q(\theta^*|\theta^{k-1})}}$$

Only the **joint distribution** $p(\mathbf{y},\theta)$ is needed - always tractable.

**For symmetric Gaussian proposals** $q(\theta^*|\theta^{k-1}) = \mathcal{N}(\theta^*|\theta^{k-1},\tau^2\mathbf{I})$:
$$q(\theta^{k-1}|\theta^*) = q(\theta^*|\theta^{k-1}) \quad \Rightarrow \quad r_k = \frac{p(\mathbf{y}, \theta^*)}{p(\mathbf{y}, \theta^{k-1})}$$

**Acceptance probability (in log domain):**
$$A_k = \min\!\left(1, \exp\left[\ln p(\mathbf{y}, \theta^*) - \ln p(\mathbf{y}, \theta^{k-1})\right]\right)$$

> **Key insights:**

> If $p_t(\theta^*) > p_t(\theta^{k-1})$: $r_k > 1$, $A_k = 1$ - always accept uphill moves

> If $p_t(\theta^*) < p_t(\theta^{k-1})$: $0 < r_k < 1$ - sometimes accept downhill (escape local modes)

> Always compute in log-domain to avoid numerical overflow/underflow

## **Quick Reference: The Algebra Toolkit**


These identities appear in almost every derivation.

| Identity | Formula |
|---|---|
| Log Gaussian (1D) | $\ln\mathcal{N}(x\|m,v) = -\frac{1}{2v}x^2 + \frac{m}{v}x + \text{const}$ |
| Log Gaussian (multivariate) | $\ln\mathcal{N}(\mathbf{x}\|\mathbf{m},\mathbf{S}) = -\frac{1}{2}\mathbf{x}^T\mathbf{S}^{-1}\mathbf{x} + \mathbf{m}^T\mathbf{S}^{-1}\mathbf{x} + \text{const}$ |
| Complete the square (1D) | $-\frac{a}{2}x^2 + bx \Rightarrow \mathcal{N}(x\|b/a, 1/a)$ |
| Complete the square (multivariate) | $-\frac{1}{2}\mathbf{x}^T\mathbf{P}\mathbf{x} + \mathbf{r}^T\mathbf{x} \Rightarrow \mathcal{N}(\mathbf{x}\|\mathbf{P}^{-1}\mathbf{r}, \mathbf{P}^{-1})$ |
