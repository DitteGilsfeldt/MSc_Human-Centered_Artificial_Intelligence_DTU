##### **Question 1.1: Determine the conditional distribution of $x_1$ given $x_2$.**

In order to determine the conditional distribution $x_1 | x_2$ we follow [Muprhy 1 section 3.1.1, where equation 3.37](../../Reading_Materials/Probabilistic_Machine_Learning__An_Introduction.pdf) shows that:


$$p(z|y) = \mathcal{N}(z|\mu_{z|y}, \Sigma_{z|y})$$

Thus, in our case, we have $z = x_1$ and $y=x_2$ so:

$$p(x_1| x_2) = \mathcal{N}(x_1|\mu_{x_1|x_2}, \Sigma_{x_1|x_2})$$

Furthermore:

$$\Sigma_{x_1|x_2}^{-1} = \Sigma_{x_1}^{-1} + A^T \Sigma_{x_2}^{-1} A$$

$$\mu_{x_1 | x_2} = \Sigma_{x_1 | x_2} [A^T \Sigma_{x_2}^{-1} (x_2 -b) + \Sigma_{x_1}^{-1} \mu_{x_1}]$$

By inserting our known values ($\Sigma_{x_1} = \Sigma_{x_2} = \Sigma$, $\mu_{x_1} = 0$, $b = 0$), we get:

$$\Sigma_{x_1|x_2}^{-1} = \Sigma^{-1} + A^T \Sigma^{-1} A$$

So:

$$\Sigma_{x_1|x_2} = (\Sigma^{-1} + A^T \Sigma^{-1} A)^{-1}$$

The mean is:

$$\mu_{x_1 | x_2} = \Sigma_{x_1 | x_2} [A^T \Sigma^{-1} x_2 + \Sigma^{-1} \cdot 0]$$

$$= \Sigma_{x_1 | x_2}\, A^T \Sigma^{-1} x_2$$

**We have thus found the conditional distribution of $x_1$ given $x_2$ with $\Sigma_{x_1|x_2} = (\Sigma^{-1} + A^T \Sigma^{-1} A)^{-1}$ and $\mu_{x_1 | x_2} = \Sigma_{x_1 | x_2}\, A^T \Sigma^{-1} x_2$.**

##### **Question 1.2: Determine the marginal distribution of $x_3$, i.e. $p(x_3)$.**

We want to find the marginal distribution $p(x_3)$. We use the **marginalisation formula**: if $y = Wz + b + \epsilon$ where $\epsilon \sim \mathcal{N}(0, \Sigma_y)$ and $z \sim \mathcal{N}(\mu_z, \Sigma_z)$, then:

$$p(y) = \mathcal{N}(y \mid W\mu_z + b,\; W\Sigma_z W^T + \Sigma_y)$$

Thus we apply this formula twice.

**Step 1:** Find $p(x_2)$ by marginalising out $x_1$. We have $z = x_1$, $y = x_2$, $W = A$, $b = 0$, $\mu_{x_1} = 0$, $\Sigma_{x_1} = \Sigma$ and $\Sigma_\epsilon = \Sigma$:

$$p(x_2) = \mathcal{N}(x_2 \mid A \cdot 0,\; A\Sigma A^T + \Sigma) = \mathcal{N}(x_2 \mid 0,\; A\Sigma A^T + \Sigma)$$

**Step 2:** Find $p(x_3)$ by marginalising out $x_2$. We have $z = x_2$, $y = x_3$, $W = A$, $b = 0$, $\mu_{x_2} = 0$, $\Sigma_{x_2} = A\Sigma A^T + \Sigma$ and $\Sigma_\epsilon = \Sigma$:

$$p(x_3) = \mathcal{N}(x_3 \mid A \cdot 0,\; A(A\Sigma A^T + \Sigma)A^T + \Sigma)$$

$$= \mathcal{N}(x_3 \mid 0,\; A^2\Sigma (A^2)^T + A\Sigma A^T + \Sigma)$$

**We have thus found the marginal distribution $p(x_3) = \mathcal{N}(x_3 \mid 0,\; A^2\Sigma (A^2)^T + A\Sigma A^T + \Sigma)$.**

**Or**

First, we compute $p(x_2)$ using the sum rule:

$$p(x_2) = \int p(x_2|x_1)p(x_1)\text{d}x_1 = \int \mathcal{N}(x_2|Ax_1, \Sigma)\mathcal{N}(x_1|0, \Sigma)\text{d}x_1 = \mathcal{N}(x_2|0, \Sigma + A\Sigma A^T)$$

and then use the sum rule a second time to compute $p(x_3)$:

$$p(x_3) = \int p(x_3|x_2)p(x_2)\text{d}x_2 = \int \mathcal{N}(x_3|Ax_2, \Sigma)\mathcal{N}(x_2|0, \Sigma + A\Sigma A^T)\text{d}x_2$$

$$= \mathcal{N}(x_3|0, \Sigma + A\Sigma A^T + AA\Sigma A^T A^T)$$

**We have thus found $p(x_3) = \mathcal{N}(x_3|0, \Sigma + A\Sigma A^T + AA\Sigma A^T A^T)$.**

##### **Question 1.3: Determine the conditional distribution of $x_3$ given $x_1$.**

We want to find the conditional distribution $p(x_3 | x_1)$. We use the sum rule:

$$p(x_3|x_1) = \int p(x_3|x_2)p(x_2|x_1)\text{d}x_2$$

$$= \int \mathcal{N}(x_3|Ax_2, \Sigma)\mathcal{N}(x_2|Ax_1, \Sigma)\text{d}x_2$$

We apply the marginalisation formula (eq. 3.38 in Murphy), where $x_2 \sim \mathcal{N}(Ax_1, \Sigma)$ plays the role of the latent variable. This gives:

$$= \mathcal{N}(x_3 | A Ax_1,\; \Sigma + A\Sigma A^T)$$

$$= \mathcal{N}(x_3 | A^2x_1,\; \Sigma + A\Sigma A^T)$$

**We have thus found the conditional distribution $p(x_3|x_1) = \mathcal{N}(x_3 | A^2 x_1, \Sigma + A\Sigma A^T)$.**
