### [Exam May 2024 — Solution Guide](02477_exam2024_guide_v2.ipynb)

**What it covers:**

**PART 1 | Linear Gaussian Systems** — Markov chain $x_1 \to x_2 \to x_3$. Q1.1: conditional $p(x_1|x_2)$ via linear Gaussian posterior formulas (precision matrix, mean from $A^T\Sigma^{-1}$). Q1.2: marginal $p(x_3)$ via two-step marginalisation using $p(y)=\mathcal{N}(y|W\mu_z+b, \Sigma_y+W\Sigma_z W^T)$. Q1.3: conditional $p(x_3|x_1)$ via Markov chain factorisation + marginalising out $x_2$.

**PART 2 | Gaussian Process Regression** — SE kernel, $\kappa=0.7$, $\ell=\frac{\sqrt{2}}{2}$, $\sigma=0.2$, $N=8$. Q2.1: Gaussian likelihood from additive noise model. Q2.2: prior predictive $p(y^*|x^*=1) = \mathcal{N}(y^*|0, \kappa^2+\sigma^2)$. Q2.3: posterior predictive at $x^*=227$ reverts to prior (kernel $\approx 0$ at large distance). Q2.4: marginal likelihood $p(y|\theta) = \mathcal{N}(y|0, K+\sigma^2 I)$, log value $\approx -6.65$. Q2.5: posterior $p(f^*|y,x^*=1)$ and predictive $p(y^*|y,x^*=1)$. Q2.6: $k_2$ not stationary/isotropic due to $c_2xx'$ term. Q2.7: prior covariance vector via $k_2$ at $x^*=-1$.

**PART 3 | Binary Classification** — quadratic feature map, MAP given. Q3.1: verifying MAP (gradient check / perturbation). Q3.2: plug-in approximation $p(y^*|y,x^*=-3)\approx\text{Ber}(y^*|\sigma(\hat{w}^T\phi^*))$. Q3.3: 90% credibility interval for $w_0$ from Gaussian approximation $S_{00}=3$. Q3.4: probit approximation $p(y^*=1)\approx\Phi(\mu_{f^*}/\sqrt{8/\pi+\sigma^2_{f^*}})$. Q3.5: expected utility + optimal decision.

**PART 4 | Variational Inference** — mean-field Gaussian, non-conjugate product likelihood $y\sim\mathcal{N}(w_1w_2,\sigma^2)$. Q4.1: entropy $H[q]=\sum_i\frac{1}{2}\ln(2\pi e v_i)$. Q4.2: expected log-likelihood via Gaussian moments $\mathbb{E}[w^2]=m^2+v$, independence. Q4.3: mean-field covariance is always 0.

**PART 5 | MCMC** — banana-shaped posterior (Rosenbrock). Q5.1: contour plot. Q5.2: Metropolis sampler, two traces. Q5.3: MC estimate of $\mathbb{E}[\sin(z_1z_2)]$. Q5.4: MC estimate of $p(z_1>z_2|\mathcal{D})$.

**Key formulas introduced:** Linear Gaussian posterior precision/mean, GP marginal likelihood, probit approximation, Gaussian moment identities for ELBO, Metropolis accept step.

### [Exam August 2024R — Solution Guide](02477_exam2024R_guide.ipynb)

**What it covers:**

**PART 1 | Multi-class Classification** — $K=3$ classes, softmax, $\phi(x)=[1,x]^T$. Q1.1: prior $p(W)=\prod_{ij}\mathcal{N}(W_{ij}|0,\alpha^{-1})$, likelihood Categorical(softmax). Q1.2: plug-in approximation via $\hat{W}_{\text{MAP}}$. Q1.3: Monte Carlo estimate by averaging softmax over 3 posterior samples. Q1.4: entropy $H=-\sum_k p_k\ln p_k$ and confidence $=\max_k p_k$. Q1.5: increasing $\alpha$ tightens prior, pushes MAP toward 0 (stronger regularisation).

**PART 2 | Gaussian Process Regression** — $k_1(x,x')=2\exp(-\frac{1}{8}\|x-x'\|^2)$, $\sigma=0.5$, $N=3$. Q2.1: identify $\kappa=\sqrt{2}$, $\ell=2$ by matching exponents. Q2.2: prior $p(f|x)=\mathcal{N}(f|0,K)$ with kernel matrix computed. Q2.3: posterior $p(f|y)=\mathcal{N}(f|m,\mathbf{K})$ via $m=K(K+\sigma^2I)^{-1}y$. Q2.4: prior variance under $k_2$ is constant $=3$ for all $x$.

**PART 3 | Mixture Prior Model** — $p(\theta)=\frac{1}{2}\mathcal{N}(\theta|-m,\tau^2I)+\frac{1}{2}\mathcal{N}(\theta|m,\tau^2I)$. Q3.1: MLE via normal equations $(X^TX)^{-1}X^Ty$. Q3.2: prior density at $[0,0]^T$ by evaluating mixture components. Q3.3: marginal likelihood by splitting integral and applying linear Gaussian marginalisation to each component. Q3.4: Bayes rule $p(\theta|y)=p(y|\theta)p(\theta)/p(y)$.

**PART 4 | Poisson GLM** — $\mu(x_n)=\exp(3+w_1x_n+w_2x_n^2)$. Q4.1: read MLE and MAP from contour plots. Q4.2: prior mean $\mathbb{E}[\mu(0)]=e^3$ (weights cancel at $x^*=0$). Q4.3: Metropolis sampler, 50% warm-up. Q4.4: $p(w_1>0|y)$ via indicator fraction. Q4.5: $p(\mu^*>7|y)$ via indicator fraction. Q4.6: 90% credibility interval from percentiles of $\mu^*$ samples.

**PART 5 | Non-linear Gaussian Model** — $y|w\sim\mathcal{N}(e^w,1)$, $w\sim\mathcal{N}(0,1)$. Q5.1: ancestral sampling for prior mean of $y$. Q5.2: log joint at $w_{\text{MAP}}$. Q5.3: Laplace approximation — Hessian $H=ye^w-2e^{2w}-1$, posterior variance $=-1/H$. Q5.4: $p(w>w_{\text{MAP}})=0.5$ by Gaussian symmetry.

**Key formulas introduced:** Softmax for multi-class, kernel hyperparameter identification by matching, mixture prior marginal likelihood, Poisson log-link GLM, Laplace approximation via second derivative.

### [Exam May 2025 — Solution Guide](02477_exam2025_guide.ipynb)

**What it covers:**

**PART 1 | Bayesian Linear Regression** — cubic polynomial, $\phi(x)=[1,x,x^2,x^3]^T$, prior $w_i\sim\mathcal{N}(0,2)$, $\sigma=1/5$, $N=5$. Q1.1: Gaussian likelihood $p(y|w,\Phi)=\mathcal{N}(y|\Phi w,\sigma^2 I)$. Q1.2: MLE via normal equations $\hat{w}=(\Phi^T\Phi)^{-1}\Phi^Ty$. Q1.3: posterior $S=(\alpha I+\frac{1}{\sigma^2}\Phi^T\Phi)^{-1}$, $m=\frac{1}{\sigma^2}S\Phi^Ty$. Q1.4: posterior predictive $p(y^*|y,x^*=3)=\mathcal{N}(y^*|m^T\phi^*, \phi^{*T}S\phi^*+\sigma^2)$.

**PART 2 | Grid Approximation** — $f(x)=e^{w_1+w_2x}$, 7-point discrete posterior. Q2.1: posterior mean $\mathbb{E}[w_1]=\sum_{w_1,w_2}w_1 q(w_1,w_2)$ and variance as weighted sum. Q2.2: posterior predictive as weighted average of Gaussian likelihoods over all grid points.

**PART 3 | Mixture Model + VI** — $p(y|\theta,x,z)=(1-z)\mathcal{N}(y|0,1)+z\mathcal{N}(y|\theta^Tx,\sigma^2)$, mean-field optimal $q_1^*$. Q3.1: $p(y_1,\theta)$ via summing over $z_1$. Q3.2: $p(y_1,z_1)$ by integrating out $\theta$ using linear Gaussian marginalisation. Q3.3: entropy $H[q_1^*]=\ln(2\pi e v)\approx3.02$. Q3.4: 95% interval via $m_2\pm1.96\sqrt{v_2}$. Q3.5: predictive mixture for $x^*=[-1,3]^T$. Q3.6: $Q_1\subset Q_2$ so KL decreases when switching to full-rank family.

**PART 4 | GP Classification** — $k_1(x,x')=5(1+\exp(-\frac{1}{4}(x-x')^2))$, Laplace approximation given. Q4.1: prior $p(f^*)=\mathcal{N}(f^*|0,10)$ from $k_1(4,4)=10$. Q4.2: prior predictive $p(y^*=0|x^*)=0.5$ by symmetry of zero-mean GP. Q4.3: posterior $\mu_{f^*|y}=k^TK^{-1}m$, $\sigma^2_{f^*|y}=k_{**}-k^TK^{-1}(K-S)K^{-1}k$. Q4.4: probit approximation $\approx0.39$. Q4.5: switching to $k_2=k_1/10$ pulls $p(y^*=1)$ toward 0.5.

**PART 5 | MCMC** — two-chain Metropolis for $p(z,x)$ where $x|z\sim\mathcal{N}(0,e^z)$, $z\sim\mathcal{N}(1,1)$, covariance $2I$. Q5.1: run chains from $(-1,0)$ and $(2,5)$, trace plots. Q5.2: $\mathbb{E}[(z-x)^2]\approx4.48$. Q5.3: $p(x<1)\approx0.66$.

**Key formulas introduced:** Grid-based discrete expectation, mean-field entropy, GP classification Laplace posterior variance formula, two-chain MCMC with combined samples.

### [Exam August 2025R — Solution Guide](02477_exam2025R_guide.ipynb)

**What it covers:**

**PART 1 | Linear Gaussian Systems** — two-step chain $x\to z\to y$ with permutation matrix $A$ ($A^TA=I$). Q1.1: $p(z|x)=\mathcal{N}(z|Ax+b,\frac{1}{2}I)$ directly from model definition. Q1.2: $p(x|z)=\mathcal{N}(x|\frac{2}{3}A^T(z-b),\frac{1}{3}I)$ via posterior precision $S^{-1}=I+2A^TA=3I$. Q1.3: $p(y)=\mathcal{N}(y|[2,1]^T,[[3.5,1.5],[1.5,2]])$ via two-step marginalisation ($p(z)$ first, then $p(y)$).

**PART 2 | Binary Classification + Laplace** — 2D logistic regression, $\lambda=1/4$, MAP given. Q2.1: plug-in $p(y^*|y,x^*=0.68)$. Q2.2: log joint $\approx-4.26$. Q2.3: gradient $=0$ at MAP by definition; Hessian $\mathcal{H}=-X^TSX-\lambda I$ where $S_{nn}=\sigma(f_n)(1-\sigma(f_n))$. Q2.4: Laplace covariance $S=-\mathcal{H}^{-1}$, probit predictive $\approx0.62$. Q2.5: replace Bernoulli with Poisson + log link for count targets.

**PART 3 | Mixture Model** — single $y$ with variance switch $s\in\{0,1\}$, $p=1/4$. Q3.1: $p(y|v)=\frac{3}{4}\mathcal{N}(y|v,1)+\frac{1}{4}\mathcal{N}(y|v,3)$ via summing over $s$. Q3.2: $p(y)=\frac{3}{4}\mathcal{N}(y|1,2)+\frac{1}{4}\mathcal{N}(y|1,4)$ via linear Gaussian marginalisation over $v$. Q3.3: $p(s=1|y)$ via Bayes rule, numerically depends on $y$.

**PART 4 | Regression + Variational Inference** — base model $y_n|w\sim\mathcal{N}(w_1+w_2x_n,\sqrt{2})$, posterior given; extended heteroscedastic model with $D=4$ mean-field approximation. Q4.1: prior $p(w_1>0)=0.5$ by symmetry; posterior $p(w_1>0|y)=1-\Phi(0.87/\sqrt{0.41})\approx0.09$. Q4.2: 80% CI for $w_2$ via $m_2\pm z_{0.90}\sqrt{S_{22}}$ ($z_{0.90}=1.282$). Q4.3: $p(f^*|y,x^*=1)=\mathcal{N}(f^*|1.26,0.90)$. Q4.4: $p(y^*|y,x^*=1)=\mathcal{N}(y^*|1.26,2.31)$. Q4.5: entropy $H[q]=\frac{1}{2}\sum_{i=1}^4\ln(2\pi e v_i)\approx2.14$. Q4.6: MC predictive $p(y^*=1|y,x^*=1)\approx0.23$ by sampling from $q(w)$ and evaluating each $\mathcal{N}$ likelihood.

**PART 5 | Non-linear Model + MCMC** — $f(x)=w_2\tanh(w_1x)$, $y=f(x)+\epsilon$, $\sigma^2=2$. Q5.1: Metropolis sampler (posterior bimodal by $\pm w$ symmetry), 10% warm-up, trace plots. Q5.2: $p(f(x^*=2)>1|y_1)\approx0.20$ via indicator MC. Q5.3: ancestral sampling for $y^*=f(x^*)+\epsilon^*$, mean $\approx0.36$, 95% interval $\approx[-2.77,3.55]$.

**Key formulas introduced:** Laplace Hessian for logistic regression, 80%/90%/95% credibility intervals with correct $z$-values ($1.282/1.645/1.96$), $D=4$ mean-field entropy, heteroscedastic model MC predictive, bimodal posterior from sign symmetry in $\tanh$.
