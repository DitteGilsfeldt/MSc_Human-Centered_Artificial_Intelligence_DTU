### 4. Estimate the parallel fraction of the program by fitting the theoretical speedup curve from Amdahl's law to your results. Nothing fancy, just play with the numbers until it looks okay. What is the estimated parallal fraction? Given this, what is your theoretical maximum speedup?

We have to use amdhal's law given as:
$$
S(p) = \frac{1}{(1 - F) + \frac{F}{p}}
$$

We also check our speedup values

```python
print("Speedup values:", speedup)
```

```python
import matplotlib.pyplot as plt
import numpy as np

p = np.array([1,2,4,8,12,16,20])
measured = speedup

f = 0.87
amdahl = 1 / ((1-f) + f/p)

plt.plot(p, measured, 'o-', label="Measured")
plt.plot(p, amdahl, '--', label="Amdahl (f=0.9)")

plt.xlabel("Processes")
plt.ylabel("Speedup")
plt.legend()
plt.grid()

plt.show()
```
