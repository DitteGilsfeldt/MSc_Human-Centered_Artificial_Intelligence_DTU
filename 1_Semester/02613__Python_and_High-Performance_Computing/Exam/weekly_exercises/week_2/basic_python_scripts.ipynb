### **Basic Python 1**

Write a Python function called listsum that sums numbers in a list.\
Input: A Python list of numbers.\
Output: A single number which is the sum of the numbers in the input list.\
Example: The input [1, 2, 3, 4] should return 10.

```python
def listsum(arr):
    sum_of_arr = 0
    for i in arr:
        sum_of_arr += i
    return sum_of_arr

listsum([1,2,3,4])
```

Write a Python function called deduplicate that removes duplicates from a list. Hint: you can use the Python builtin \
Input: A Python list which may contain duplicate elements.\
Output: A Python list containing the unique elements of the input list.\
Example: The input [1, 2, 3, 3, 2, 2, 4] should return [1, 2, 3, 4].



```python
def deduplicate(arr):
    arr_set = set(arr)
    return list(arr_set)

deduplicate([1,1,1,2])
```

### **Basic Python 3**

Write a Python function called sorttuples that sorts a list of tuples according to their last element. Hint: you may use the builtin function sorted. Look at the key parameter.\
Input: A Python list of tuples.\
Output: The input list with elements sorted according to the last element in each tuple.\
Example: The input [(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)] should return [(2, 1), (1, 2), (2, 3), (4, 4), (2, 5)].


```python
def sorttuples(arr):
    return sorted(arr, key = lambda x: x[-1])

sorttuples([(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)])
```

### **Basic Python 4**

Write a Python function squarecubes which recieves a Python list of numbers and returns two lists containing, respectively, the squares and cubes of each number.\
Input: A Python list of numbers.\
Output: A tuple of two Python lists. The first list contains the squares of the numbers in the input list. The second list contains the cubes.\
Example: The input [1, 2, 3, 4] should return ([1, 4, 9, 16], [1, 8, 27, 64]).

```python
def squarecubes(arr):
    cubes = [i**2 for i in arr]
    squares = [i**3 for i in arr]
    return cubes, squares

squarecubes([1,2,3,4])
```

### **Basic Python 7**


Write a Python class Student, that has the student name and a list of courses the student is attending as attributes. These should given as argument to the class constructor. The class should also have a method attends that receives a course name and returns True if the student is attending that course and False if not.\
Constructor input: The first argument is a string with the student's name. The second is a list of course names, where each course name is a string.\
attends input: A string with a course name.\
attends output: True if the student attends that course, i.e., it is in the list of courses the student attends. False if not.
Example: If we create a student as
s = Student('X', ['01005', '02613']), we expect that
s.attends('01005') and s.attends('02613') returns True and
s.attends('02510') returns False.

```python
class Student:
    def __init__(self, name, courses):
        self.courses = courses
        self.name = name

    def attends(self, course):
        if course in self.courses:
            return True
        else:
            return False
        
s = Student("X", ['01005', '02613'])
s.attends("0105")
```

### **Basic Python 8**


Write a Python function coursestudents that a list of Student classes as well as a course name. It must return a new list containing the names of the students that attends that course.\
Input: The first argument is a Python list of Students. The second is a string specifying a course name.\
Output: A list containing the names of the students that attends the given course.\
Example: Given the list of Students:
students = [Student('A', ['01005']), Student('B', ['02613']), <br>Student('C', ['01005', '02613'])], we expect that
coursestudents(students, '02613') returns ['B', 'C'].

```python
def coursestudents(students, course):
    students_attending_course = []
    for student in students:
        if course in student.courses:
            students_attending_course.append(student.name)
    return students_attending_course
        
coursestudents([Student('A', ['01005']), Student('B', ['02613']), Student('C', ['01005', '02613'])], "02613")
```

### **Basic Numpy 1**


Write a Python function magnitude, which takes an $\eta$-dimensional vector stored as a NumPy array as input and return its magnitude, i.e., norm.\
Input: An $\eta$-dimensional vector stored as a NumPy array.\
Output: A number giving the magnitude of the input vector.\
Example: For the input [1, 1, 3, 3, 4] we expect the output 6.



```python
def magnitude(x):
    lst = []
    for i in x:
        i = i**2
        lst.append(i)
    sum_of_lst = sum(lst)
    return sum_of_lst**(1/2)

magnitude([1, 1, 3, 3, 4])
```
