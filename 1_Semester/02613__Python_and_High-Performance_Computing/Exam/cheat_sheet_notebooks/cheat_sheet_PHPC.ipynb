### **Glossary of concepts**


</div>


##### **Hardware & Architecture**



**CPU (Central Processing Unit)** - The main processor of a computer. Has a small number of very fast, flexible cores (typically 4–64). Each core is independently fast and good at sequential logic, branching, and general computation. Fundamentally serial with parallelism added on top.

**GPU (Graphics Processing Unit)** - A co-processor attached to the CPU via the PCIe bus. Has thousands of small, slow cores designed for massive parallelism. Each core is slow and simple but there are so many that throughput is enormous. Fundamentally parallel with serial logic added on top. Also called the **device** (CPU = **host**).

**Core** - One independent processing unit inside a CPU or GPU. CPU cores are fast and flexible; GPU cores are slow and simple.

**RAM (Random Access Memory)** - The main working memory of the computer (~8–256 GB). Much slower than CPU caches (~200 CPU cycles to access). Temporary - lost when the computer is turned off.

**SSD / HDD** - Permanent storage on disk. Orders of magnitude slower than RAM (microseconds for SSD, milliseconds for HDD). Data persists after shutdown.

**PCIe bus** - The physical connection between the CPU and GPU. Data transferred over it takes ~0.6 s for a 10 GB array - this is the GPU transfer overhead.

**NUMA (Non-Uniform Memory Access)** - A multi-socket server has two physical CPU chips, each with its own RAM bank. Cores on socket 0 access socket 0's RAM fast (local), but accessing socket 1's RAM is slower (remote). Performance drops when crossing sockets.


##### **Memory Hierarchy**

**Memory hierarchy** - The chain of memory levels from fastest/smallest to slowest/largest: Registers → L1 Cache → L2 Cache → L3 Cache → RAM → SSD. When the CPU needs data it checks each level in order - if not found (cache miss), it goes to the next level down.

**Register** - Tiny storage directly inside the CPU core. Instant access. Holds the values currently being computed.

**Cache (L1/L2/L3)** - Small, fast memory between the CPU and RAM. L1 is fastest (~4 CPU cycles), L3 is slowest (~60 cycles) but largest. Automatically managed by hardware. A **cache hit** means the data was already there; a **cache miss** means it had to be fetched from the next level.

**Cache line** - The unit of transfer between cache levels. Always 64 bytes (8 float64 values). When you access any element, the CPU loads the whole 64-byte block containing it. This is why accessing nearby elements is fast - they were already loaded.

**Spatial locality** - The principle that if you access memory address N, you will likely soon need N+1, N+2, etc. Row-major access exploits this; column access breaks it.

**Temporal locality** - The principle that if you access data now, you will likely access it again soon. Loops that reuse the same values keep them in cache.

**Cache miss** - The requested data is not in cache. The CPU stalls and waits for it to be fetched from the next level. With random access on large data, almost every access is a cache miss, forcing fetches from RAM (200+ cycles each).

**Row-major (C-order)** - How NumPy stores arrays by default. The last index varies fastest in memory. For a 2D array `a[i, j]`, elements in the same row are adjacent: `a[0,0], a[0,1], a[0,2]...`. Row access is cache-friendly; column access is not.

**Stride** - How many bytes to jump in memory to move one step along a 
given array dimension. For a `(M, N)` float64 array, stride along axis 0 
(rows) = `N*8`, stride along axis 1 (cols) = `8`.

**Reduction operator** - not in the glossary. Worth adding:
> An operator that collapses an array to a single value (sum, product, AND). Must be both **commutative** and **associative** to be used in a parallel tree reduction framework.

**Tree reduction** - not defined. Worth one line:
> A parallel reduction algorithm that halves the active range each round, taking $\log_2(N)$ rounds instead of $N-1$ sequential steps.

**`numactl --interleave=all`** - NUMA is defined but the fix isn't in the glossary. Just one line:
> Distributes memory pages round-robin across all NUMA nodes so neither socket has all-remote accesses. Fixes the performance drop beyond ~24 cores.


### **Parallelism Concepts**



**Serial** - Code that runs one step at a time, in order. Step 2 cannot start until step 1 is complete. All standard Python code is serial by default.

**Parallel** - Multiple tasks executing at the exact same instant on different cores or processors. True parallelism requires multiple physical cores.

**Concurrent** - Tasks that overlap in time but may not run at exactly the same instant. One task may wait (e.g. for I/O) while another runs. Concurrency is a broader concept - parallelism is a subset of concurrency.

**Parallelisable** - A loop or computation is parallelisable if and only if **no iteration depends on the output of another iteration**. The one-question test: could you run iteration 50 without having run iterations 1–49? If yes → parallelisable.

**Sequential dependency** - When iteration $i$ uses the result of iteration $i-1$ as input. Example: `x = f(x)` inside a loop. These cannot be parallelised.

**Embarrassingly parallel** - A problem that can be split into completely independent subtasks with zero communication needed between them. Example: applying the same function to each image in a dataset. Gives the best possible speedups.

**Speedup** $S(p)$ - How much faster the parallel version is compared to serial: $S(p) = T(1) / T(p)$. Perfect speedup would be $S(p) = p$ (using $p$ cores makes it $p\times$ faster). Reality is always less due to overhead and serial fractions.

**Parallel fraction** $F$ - The fraction of total runtime that can be parallelised. The rest ($1-F$, the serial fraction $B$) is irreducible overhead that bounds maximum speedup.

**Amdahl's law** - The theoretical maximum speedup given parallel fraction $F$ and $p$ processors: $S(p) = 1 / ((1-F) + F/p)$. As $p \to \infty$, max speedup $= 1/(1-F)$. The serial fraction is the hard ceiling.

**Serial fraction** $B = 1-F$ - The part of the program that must run sequentially regardless of how many cores you add. Even 1% serial → max speedup of 100×.

**Efficiency** $E(p) = S(p)/p$ - Fraction of processors actually doing useful work. Decreases as $p$ grows due to overhead and the serial fraction.

**Load balancing** - Distributing work evenly across workers so no worker sits idle. Important when tasks have variable runtimes.

**Static scheduling** - Divide all work into equal chunks upfront and assign one chunk per worker. Fast setup, but if chunks have variable runtime, fast workers idle. Good only when all tasks take the same time.

**Dynamic scheduling** - Create many small chunks. Workers grab the next chunk whenever they finish the previous one. Automatically balances load. Better for variable runtimes.

**Overhead** - The extra time spent on parallelism infrastructure: spawning processes, pickling data, sending it between processes. If overhead > work per task, parallelism makes things slower.

**Chunking** - Grouping many small tasks into a larger batch so each worker gets meaningful work before going back to the queue. Reduces overhead from too many IPC round-trips.

#### **CPU Bound vs I/O Bound vs GPU**


**CPU-bound** - The bottleneck is the CPU doing computation. Adding more cores or a faster CPU speeds things up. Signals: simulations, numerical calculations, matrix operations, running for minutes/hours.

**I/O-bound** - The bottleneck is waiting for data: disk reads, network downloads, database queries. The CPU is mostly idle waiting. Signals: file loading, web requests, any mention of "waiting" or "reading". Threading (not multiprocessing) is the right fix because threads can overlap their waiting time.

**GPU-suited** - The bottleneck is a large number of independent operations on array data. The GPU's thousands of cores run them all simultaneously. Signals: "each element is updated independently", "no dependencies between elements", large arrays, fixed arithmetic per element.

**GPU not suited** - Sequential logic, variable-length loops per element (like `while height > 0`), Python dictionaries/objects, code that cannot be modified to use `@cuda.jit`.

#### **Threading vs Multiprocessing**


**Thread** - A lightweight execution unit that shares memory with other threads in the same process. In Python, threads are blocked from true CPU parallelism by the GIL.

**Process** - A fully independent Python interpreter with its own memory space and GIL. Processes don't share memory by default. `multiprocessing` spawns separate processes - they can truly run in parallel.

**GIL (Global Interpreter Lock)** - A lock in CPython that allows only one thread to execute Python bytecode at a time. Prevents race conditions on reference counts but also prevents CPU-bound threads from running in parallel. **Threading cannot speed up CPU-bound Python code.**

**GIL exception** - Some operations release the GIL at the C level, allowing true thread parallelism: NumPy BLAS operations (`np.matmul`, `np.dot`), file I/O, network calls. `ThreadPool` works for these.

**Race condition** - When two threads/processes read and write the same shared variable simultaneously, producing unpredictable results. Example: both read `a=2`, both add to it, one writes `4`, the other also writes `5` - the final value depends on who writes last.

**Deadlock** - Two threads each wait for the other to release a lock, so both wait forever and nothing progresses.

**Pickle** - Python's serialisation format for sending objects between processes. `multiprocessing` pickles data automatically. Large objects are slow to pickle - prefer shared memory (`mp.RawArray`) for large arrays.

**Shared memory** (`mp.RawArray`) - A block of memory accessible to all worker processes without copying. Required for efficient parallel reduction where workers need to read/write the same large array.


##### **GPU Computing**

**Kernel** - A function decorated with `@cuda.jit` that runs on the GPU. Every thread executes the same kernel code but on a different data element. Kernels cannot return values - results must be written to an output array.

**Thread** (GPU) - The smallest unit of GPU execution. Runs the kernel on one output element. Thousands execute simultaneously.

**Thread block** - A group of threads that run on the same Streaming Multiprocessor (SM) and share L1 cache and shared memory. Specified as `threads_per_block` when launching a kernel.

**Grid** - The collection of all thread blocks for a kernel launch. `cuda.grid(1)` gives a thread its global index across the entire grid.

**Warp** - 32 threads that execute the exact same instruction simultaneously (SIMT). The GPU's hardware unit of execution. You never write warps directly - the hardware groups threads into warps automatically based on thread ID.

**Warp divergence** - When threads in the same warp take different branches (`if/else` based on thread index), the GPU executes both branches serially with half the threads inactive each time. Up to 32× slowdown. Avoid branching inside kernels.

**Coalesced access** - When 32 warp threads read 32 consecutive memory addresses simultaneously, the GPU satisfies all reads in one memory transaction. Non-coalesced (scattered) access requires up to 32 separate transactions - 32× slower.

**SIMT (Single Instruction, Multiple Threads)** - The GPU execution model. All 32 threads in a warp execute the same instruction at the same time, each on different data.

**HtoD (Host to Device)** - Transferring data from CPU memory to GPU memory. Automatic when passing NumPy arrays to a Numba kernel. Expensive - minimise with manual transfer using `cuda.to_device()`.

**DtoH (Device to Host)** - Transferring data from GPU memory back to CPU memory. Automatic after a Numba kernel returns. Use `d_arr.copy_to_host()` for manual control.

**Shared memory** (GPU) - A small, fast memory (~100 KB) local to one thread block, roughly L1 cache speed. Manually managed. Useful when threads in a block access the same data repeatedly or in non-coalesced patterns. Always call `cuda.syncthreads()` after writing shared memory before reading it.

**Arithmetic intensity** - FLOPs per byte transferred from memory. Low intensity (e.g. vector addition: 1 FLOP per 16 bytes) means the GPU is memory-bound and transfer dominates. High intensity (e.g. matmul: ~N/2 FLOPs per byte) means the GPU is compute-bound and transfer cost amortises.

**Pinned memory** - CPU memory locked in RAM (cannot be swapped to disk). Allows faster HtoD/DtoH transfers because the GPU's DMA engine can access it directly. Created with `cuda.pinned_array()` or `cuda.pinned()`.

**GPU overhead trap** - GPU benchmarks at small iteration counts look slow because fixed transfer cost dominates. Always compute per-iteration cost as $\Delta T / \Delta N$ from two measurements. If GPU per-iteration cost < CPU per-iteration cost, the GPU wins at real workload scale.

##### **Profiling**

**Profiling** - Measuring where a program spends its time. Never optimise without profiling first - the bottleneck is almost never where you expect.

**`cProfile`** - Function-level profiler. Low overhead. Use it first to find which function is slow. Key columns: `tottime` (time inside the function, excluding sub-calls) and `cumtime` (total including sub-calls).

**`kernprof` / `line_profiler`** - Line-level profiler. High overhead. Use only on the confirmed bottleneck function. Reports time per line in µs and number of hits.

**`nsys` (Nsight Systems)** - NVIDIA's GPU profiler. Captures HtoD/DtoH transfers and kernel execution times. Used to diagnose whether a GPU program is transfer-bound or compute-bound. So nsys is basically fixed overhead = HtoD + DtoH time, per-iter = kernel time / N iterations
cuLaunchKernel thousands of times = Python loop calling small CuPy ops → fuse into one call

**Wall-clock time** - The actual time you waited. Measured with `perf_counter`. The most honest metric.

**CPU time** - Total CPU seconds consumed across all cores. For parallel code, CPU time > wall time (multiple cores running). For I/O-bound code, CPU time < wall time (CPU was idle waiting).

**FLOP/s** - Floating point operations per second. Measures raw compute speed, independent of how much work was done. A drop in FLOP/s means the CPU is waiting for data, not computing.


##### **Data Storage**

**Chunking** (Zarr/pandas) - Processing data in fixed-size pieces so only one chunk lives in RAM at a time. Allows processing datasets larger than RAM.

**Chunk shape** - For Zarr arrays, the shape of each stored chunk. Must match the access pattern: wide chunks for row access, tall chunks for column access, square chunks for neighbourhood access.

**Memory mapping** (`np.memmap`) - Makes a disk file appear as a NumPy array in memory. The OS loads only the pages actually accessed. Allows arrays larger than RAM without loading the whole file.

**Zarr** - A chunked, compressed array format stored as a directory of chunk files. Each chunk is a separate file → multiple processes can write simultaneously. Better compression than raw binary. So a Zarr array is stored on disk in rectangular blocks. When we access any element inside a block, the entire block gets loaded into memory. 

**Parquet** - A binary columnar file format for tabular data. ~10× faster to read than CSV and much smaller on disk because numbers are stored as raw bytes (not ASCII text) and compressed column-by-column.

**Category dtype** - Pandas encoding for columns with few unique values. Stores unique values once in a lookup table and the column as integer indices. Huge memory savings when `#unique << #rows`.


##### **HPC / Job Scheduling**

**Scheduler (LSF)** - The software on the HPC cluster that decides when and where each job runs based on available resources. You submit jobs to a queue; the scheduler runs them when resources are free.

**Wall time** (`-W`) - The maximum time a job is allowed to run. If it exceeds this, the scheduler kills it. Format: `HH:MM` with colon, or bare minutes without.

**Per-core memory** (`rusage[mem=XGB]`) - Memory requested **per CPU core**, not total. Total memory = per-core × number of cores. Easy exam trap.

**Job array** - A set of many identical jobs submitted at once, each with a different `$LSB_JOBINDEX`. Used for embarrassingly parallel workloads where the same script runs on different data.

**Job dependency** (`-w done(jobname)`) - Makes a job wait in PEND state until another job reaches a specified state. `done()` requires all elements to complete successfully; `ended()` fires on success or failure.

**`span[hosts=1]`** - Forces all requested cores onto the same physical node. Required for any code using shared memory (`mp.RawArray`) or multiprocessing, since shared memory is per-machine.
