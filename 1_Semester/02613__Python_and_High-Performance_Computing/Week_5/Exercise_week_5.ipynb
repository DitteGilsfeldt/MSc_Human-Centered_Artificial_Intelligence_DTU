## 1. Amdahl's Law

<img src="Images/2026-03-08-12-09-09.png" width="650">

Amdahl’s Law says that if a fraction (f) of a program can be parallelized, then the speedup with (p) processors is

$$
S(p)=\frac{1}{(1-f)+\frac{f}{p}}
$$

where:

* (f) = parallel fraction
* (1-f) = serial fraction
* (p) = number of processors

In this exercise, the task has two parts:

* Subtask 1: 20 seconds, cannot be parallelized
* Subtask 2: 100 seconds, can be parallelized

So the total sequential runtime is

$$
T_1 = 20 + 100 = 120 \text{ seconds}
$$

### 1. What is the parallel fraction of the task consisting of the two subtask? 

The parallel part is the 100-second subtask, so

$$
f = \frac{100}{120} = \frac{5}{6} \approx 0.8333
$$

So the parallel fraction is

$$
\boxed{f = \frac{5}{6} \approx 0.8333}
$$

The serial fraction is

$$
1-f=\frac{20}{120}=\frac{1}{6}\approx 0.1667
$$

```python
# --- Given values ---
serial_time = 20        # seconds (cannot be parallelized)
parallel_time = 100     # seconds (can be parallelized)
total_time = serial_time + parallel_time

# --- 1. Parallel fraction ---
f = parallel_time / total_time
print("Parallel fraction (f):", f)

# --- Amdahl's Law function ---
def amdahl_speedup(f, p):
    return 1 / ((1 - f) + (f / p))
```

### 2. What is the theoretical speedup for $p=10$ processors? 

Use Amdahl’s Law:

$$
S(10)=\frac{1}{(1-f)+\frac{f}{10}}
$$

Insert (f=\frac{5}{6}):

$$
S(10)=\frac{1}{\frac{1}{6}+\frac{5}{6\cdot 10}}
$$

$$
S(10)=\frac{1}{\frac{1}{6}+\frac{5}{60}}
=\frac{1}{\frac{10}{60}+\frac{5}{60}}
=\frac{1}{\frac{15}{60}}
=\frac{60}{15}=4
$$

So the theoretical speedup is

$$
\boxed{4}
$$

That means the program becomes 4 times faster with 10 processors.

You can also check this by runtime:

* Serial part stays 20 s
* Parallel part becomes (100/10 = 10) s

So total runtime with 10 processors is

$$
20 + 10 = 30 \text{ s}
$$

and speedup is

$$
\frac{120}{30}=4
$$

```python
p = 10
speedup_10 = amdahl_speedup(f, p)
print("Speedup with 10 processors:", speedup_10)
```

### 3. What is the theoretical maximum speedup? 

The maximum speedup happens when the number of processors goes to infinity, so the parallel part takes almost no time:

$$
S_{\max}=\frac{1}{1-f}
$$

Since (1-f=\frac{1}{6}),

$$
S_{\max}=\frac{1}{1/6}=6
$$

So the theoretical maximum speedup is

$$
\boxed{6}
$$

This is the key idea of Amdahl’s Law: even with infinitely many processors, the serial part still limits the total speedup.

```python
max_speedup = 1 / (1 - f)
print("Maximum theoretical speedup:", max_speedup)
```

### 4. Imagine we currently have 4 processor available for the parallization. To improve performance, we have two options:
1. Optimize the first subtask so we can retrieve the data from the file in 5 second instead og 20. 
2. Purchase more processors so we have 8 processors available instead of 4.

which option will provide the shortest runtime? 

We compare the two options.

Current situation with 4 processors

Runtime is:

* Serial part: 20 s
* Parallel part: $100/4 = 25$ s

So current runtime is

$$
20 + 25 = 45 \text{ s}
$$

Now compare the two improvements.

Option 1: Optimize the first subtask from 20 s to 5 s

Now runtime becomes:

* First subtask: 5 s
* Second subtask with 4 processors: $100/4 = 25$ s

Total:

$$
5 + 25 = 30 \text{ s}
$$

So option 1 gives

$$
\boxed{30 \text{ s}}
$$

Option 2: Increase processors from 4 to 8

Now runtime becomes:

* First subtask: 20 s
* Second subtask with 8 processors: $100/8 = 12.5$ s

Total:

$$
20 + 12.5 = 32.5 \text{ s}
$$

So option 2 gives

$$
\boxed{32.5 \text{ s}}
$$

Conclusion

* Option 1 runtime: (30) s
* Option 2 runtime: (32.5) s

So the shortest runtime comes from

$$
\boxed{\text{Option 1: optimize the first subtask}}
$$

Why?

This is exactly what Amdahl’s Law is about: improving the serial part can be more valuable than adding more processors, because the serial part is the bottleneck that cannot be parallelized.

Even though buying more processors helps the 100-second part, it does nothing to reduce the fixed 20-second serial part. Reducing that serial part from 20 to 5 seconds cuts a big bottleneck.

Here is a clean version you can hand in:

The total sequential runtime is (120) seconds, consisting of $20$ seconds serial work and $100$ seconds parallel work. Therefore, the parallel fraction is $f=\frac{100}{120}=\frac{5}{6}\approx 0.8333$. Using Amdahl’s Law, the theoretical speedup for $p=10$ processors is $S(10)=\frac{1}{(1-f)+f/10}=\frac{1}{1/6+ (5/6)/10}=4$. The theoretical maximum speedup is reached as $p\to\infty$, giving $S_{\max}=\frac{1}{1-f}=\frac{1}{1/6}=6$. With 4 processors, the runtime is $20+100/4=45$ seconds. If the first subtask is optimized from $20$ to $5$ seconds, the runtime becomes $5+100/4=30$ seconds. If instead the number of processors is increased from $4$ to $8$, the runtime becomes $20+100/8=32.5$ seconds. Therefore, optimizing the first subtask gives the shortest runtime.

```python
# Current situation with 4 processors
p_current = 4
current_runtime = serial_time + parallel_time / p_current
print("Current runtime with 4 processors:", current_runtime)

# Option 1: optimize first task (20s -> 5s)
optimized_serial = 5
runtime_option1 = optimized_serial + parallel_time / p_current
print("Runtime if first task optimized:", runtime_option1)

# Option 2: increase processors (4 -> 8)
p_new = 8
runtime_option2 = serial_time + parallel_time / p_new
print("Runtime with 8 processors:", runtime_option2)

# Which option is better?
if runtime_option1 < runtime_option2:
    print("Best option: Optimize the first subtask")
else:
    print("Best option: Buy more processors")
```
